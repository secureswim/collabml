```python
import collabml

collabml.configure(
    api_url="https://YOUR-COLLABML-SERVER.example.com",
    token="YOUR_PRIVATE_TOKEN",
)
```

```python
# Run once. If the project already exists, skip this cell.
collabml.create_project(
    name="mnist-classifier",
    description="Explore small neural networks on MNIST",
    primary_metric="val_accuracy",
    goal="maximize",
)
```

```python
run = collabml.start(
    project="mnist-classifier",
    title="Dense baseline",
    hypothesis="A single hidden layer should exceed 95% validation accuracy",
    params={"hidden_units": 128, "learning_rate": 0.001, "epochs": 5},
    notebook_url="PASTE_THIS_COLAB_NOTEBOOK_URL",
)
print(run.id, run.url)
```

```python
# Train your real model here, then record the meaningful outcome.
run.log({"val_accuracy": 0.963, "val_loss": 0.121})
run.complete("The baseline clears 95%; try dropout to reduce validation loss.")
run.url
```

```python
fork = collabml.fork(
    run.id,
    title="Add dropout",
    hypothesis="Dropout should reduce validation loss without hurting accuracy",
    changes={"dropout": 0.25},
)
fork.log({"val_accuracy": 0.968, "val_loss": 0.104})
fork.complete("Dropout improved both tracked metrics.")
fork.url
```
